#### Nettoyer les données

```python
# Identifier les valeurs manquantes
apple.isnull().sum()
```

```python
apple.isnull().mean()
```

```python
# Verification des doublants
apple.duplicated().sum()
```

```python
# traiter les valeurs manquantes
## Supprimer les valeurs manquantes
apple_clean =apple.dropna()
```

```python
apple_clean.head()
```

```python
# Supprimer les doublants
apple_clean = apple_clean.drop_duplicates()
```

#### 3- Gérer les dates et les fréquences

```python
apple_clean.index
```

```python
apple_clean.index = pd.to_datetime(apple_clean.index) #Convertir en format date
```

```python
type(apple_clean.index)
```

```python
# transformer les valeurs en fréquence journalière 
weekly_date = apple_clean.resample("w").mean() # Transformer des données journalières en données hebdomadaires 
```

```python
weekly_date.head()
```

```python
monthly_date = apple_clean.resample("ME").mean()# Transformer des données journalières en données mensuelles 
```

```python
monthly_date.head()
```

```python
quartely_date = apple_clean.resample("QE").mean()# Transformer des données journalières en données trimestrielles
quartely_date.head()
```

```python
yearly_date = apple_clean.resample("YE").mean()# Transformer des données journalières en données annuelles
yearly_date.head()
```

### Echantillonner les données

```python
# uniformiser
apple_clean["Volume"] = apple_clean["Volume"]/1000
apple_clean["Volume"].describe()
```

```python
# prendre un échantillon de 20% aléatoirement pour tester des annalyses
sample = apple_clean.sample(frac=0.2)
sample.describe()
```

```python
# Por fixer les aléa
sample = apple_clean.sample(frac=0.2, random_state=255)
sample.describe()
```

```python
# On peut faire l'échantillonnage en fonction du nombre d'observation
# Echantillon
sample = apple_clean.sample(n=200, random_state=255)
sample.describe()
```
